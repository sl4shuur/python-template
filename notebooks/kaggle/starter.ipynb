# Kaggle starter

Attach a wheelhouse Dataset. Offline-safe install; set the competition data path.

```python
import glob, pathlib, subprocess, sys
# Attach your wheelhouse Dataset on the right sidebar (Data > Add Input).
wheels = glob.glob("/kaggle/input/**/kaggle_starter-*.whl", recursive=True)
if not wheels:
    raise FileNotFoundError("Attach the kaggle-starter-wheelhouse Dataset containing the project wheel")
wheelhouse = pathlib.Path(wheels[0]).parent
requirements = wheelhouse / "requirements.txt"
if not requirements.exists():
    raise FileNotFoundError(f"Missing wheelhouse requirements: {requirements}")
subprocess.run([sys.executable, "-m", "pip", "install", "--no-index", "--find-links", str(wheelhouse), "-r", str(requirements)], check=True)
```
